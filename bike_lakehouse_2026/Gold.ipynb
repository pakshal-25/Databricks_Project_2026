#Business Transformation

```python
query = """
    SELECT 
        customer_id,
        customer_key,
        first_name,
        last_name
    FROM
        silver.crm_customers 
"""
df = spark.sql(query)
display(df)
```

#Write it to Gold Table


```python
(
    df.write
    .mode("overwrite")
    .format("delta")
    .saveAsTable("gold.dim_customers")
)
```
