# **_Transformation_**

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY pn.start_date, pn.product_number) AS product_key,
    pn.product_id,
    pn.product_number,
    pn.product_name,
    pn.category_id,
    pc.category,
    pc.subcategory,
    pc.maintenance_flag,
    pn.product_line,
    pn.start_date
FROM workspace.silver.crm_product pn
LEFT JOIN workspace.silver.erp_px_cat_g1v2 pc
    ON pn.category_id = pc.category_id
"""
df = spark.sql(query)
```

# **_Writing Gold Table_**

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_product")
```
