```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder.appName("EcomDataPipeline").getOrCreate()
```

```python
user_df = spark.read.format("parquet")\
            .option('header','true') \
                .option('inferSchema','true')\
                    .load('/Volumes/workspace/bronze_ecom/ecom_source/ecom_data/chunk1.parquet')
user_df.display()
```

```python
user_df.write.format("delta")\
    .mode("overwrite")\
        .saveAsTable("bronze_ecom.users")
```

```python
buyers_df = spark.read.format("parquet")\
    .option("header","true")\
        .option('inferSchema','true')\
            .load('/Volumes/workspace/bronze_ecom/ecom_source/ecom_data/Buyers-repartition-by-country.parquet')
buyers_df.display()
```

```python
buyers_df.write.format('delta')\
    .mode('overwrite')\
        .saveAsTable('bronze_ecom.buyers')
```

```python
sellers_df = spark.read.format("parquet")\
    .option("header","true")\
        .option('inferSchema','true')\
            .load("/Volumes/workspace/bronze_ecom/ecom_source/ecom_data/Comparison-of-Sellers-by-Gender-and-Country.parquet")
sellers_df.display()
```

```python
sellers_df.write.format('delta')\
    .mode('overwrite')\
        .saveAsTable('bronze_ecom.sellers')
```

```python
countries_df = spark.read.format('parquet')\
    .option('header','true')\
        .option('inferSchmea','true')\
            .load('/Volumes/workspace/bronze_ecom/ecom_source/ecom_data/Countries-with-Top-Sellers-(Fashion-C2C).parquet')
countries_df.display()
```

```python
countries_df.write.format('delta')\
    .mode('overwrite')\
        .saveAsTable('bronze_ecom.countries')
```
